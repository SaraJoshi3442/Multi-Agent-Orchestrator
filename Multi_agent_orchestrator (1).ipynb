```python
def research_agent(task):
  print("Research agent received:",task)
  result = {
      "topic": task,
      "summary":"This research is about: "+task ,
      "references":["reference 1","reference 2"]
  }
  return result
```

```python
research_result = research_agent("Henry Winter")
print(research_result)
```

```python
def writing_agent(research_agent):
  topic = research_agent["topic"]
  summary = research_agent['summary']
  references = research_agent['references']

  result= "report about "+topic+"\n"+summary+"\n"+str(references)
  return result
```

```python
def orchestrator(task):
  research_result=research_agent(task)
  writing_result = writing_agent(research_result)
  print(writing_result)
```

```python
def research_agent(task):
  prompt = "Research this topic and explain it clearly: "+task
  response = client.models.generate_content(
      model="gemini-3.5-flash-lite",
      contents=prompt
  )
  return response.text
```

```python
result = research_agent("A Good Girl's Guide To Murder")
print(result)
```

```python
def writing_agent(result):
    prompt = f"""
    You are a professional writing agent.

    Convert the following research into a well-structured report.
    Include a title, headings, and a conclusion.

    Research:
    {result}
    """

    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )

    return response.text
```

```python
report = writing_agent(result)
print(report)
```

```python
def fact_checker_agent(report):
    prompt = f"""
    You are a careful fact-checking assistant.

    Review the report below.

    Your tasks:
    1. Identify factual claims that may be incorrect.
    2. Identify claims that need evidence or verification.
    3. Check whether names, dates, titles, and statistics
       are internally consistent.
    4. Do not assume a claim is true just because it
       sounds convincing.
    5. Do not invent corrections or sources.
    6. If you cannot verify a claim, label it
       as "Needs verification".
    7. Return a corrected version of the report,
       followed by a short list of important corrections
       and claims that still need verification.

    Report:
    {report}
    """

    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )

    return response.text
```

```python
fact = fact_checker_agent(report)
print(fact)
```

```python
def orchestrator(task):
    task_lower = task.lower()

    if 'research' in task_lower:
        return research_agent(task)

    elif 'report' in task_lower or 'write' in task_lower:
        research = research_agent(task)

        report = writing_agent(research)

        checked_report = fact_checker_agent(report)

        return checked_report

    else:
        return "I don't know which agent to use"
```

```python
result = orchestrator(
    "Write a report about A Good Girl's Guide to Murder"
)
print(result)
```
