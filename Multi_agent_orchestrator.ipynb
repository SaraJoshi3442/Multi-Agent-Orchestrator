make the agent return strucutred data

```python
def research_agent(task):
  print("Research agent received:",task)
  result = {
      "topic": task,
      "summary":"This research is about: "+task ,
      "references":["reference 1","reference 2"]
  }
  return result
```

```python
research_result = research_agent("Henry Winter")
print(research_result)
```

```python
def writing_agent(research_agent):
  topic = research_agent["topic"]
  summary = research_agent['summary']
  references = research_agent['references']

  result= "report about "+topic+"\n"+summary+"\n"+str(references)
  return result
```

```python
def orchestrator(task):
  research_result=research_agent(task)
  writing_result = writing_agent(research_result)
  print(writing_result)
```

```python
orchestrator("Henry Winter")
```

```python
def research_agent(task):
  response = client.models.generate_content(
      model="gemini-3.5-flash-lite",
      contents="Research the following topic: "+task
  )
  return response.text
```

```python
result = research_agent("The Secret History by Donna Tartt")
print(result)
```

```python
def writing_agent(result):
  response = client.models.generate_content(
      model="gemini-3.5-flash-lite",
      contents="Write a report using the following research: "+result
  )
  return response.text
```

```python
report = writing_agent(result)
print(report)
```

```python
def orchestrator(task):
  response = research_agent(task)
  report = writing_agent(response)
  print(report)
```

```python
result = orchestrator("Kristian Hannah Books")
print(result)
```

```python
def orchestrator(task):
  task_lower = task.lower()
  if 'research' in task_lower:
    response = research_agent(task)
    report = writing_agent(response)
    print(report)
  elif 'report' in task_lower or 'write' in task_lower:
    research = research_agent(task)
    return writing_agent(research)
  else:
    return "I don't know which agent to use"
```

```python
result = orchestrator("write a report about the book lessons in chemistry")
print(result)
```
