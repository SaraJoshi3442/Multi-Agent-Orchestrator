```python
def writing_agent(result):
    prompt = f"""
    You are a professional writing agent.

    Convert the following research into a well-structured report.
    Include a title, headings, and a conclusion.

    Research:
    {result}
    """

    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )

    return response.text
```

```python
report = writing_agent(result)
print(report)
```

```python
def fact_checker_agent(report):
    prompt = f"""
    You are a careful fact-checking assistant.

    Review the report below.

    Your tasks:
    1. Identify factual claims that may be incorrect.
    2. Identify claims that need evidence or verification.
    3. Check whether names, dates, titles, and statistics
       are internally consistent.
    4. Do not assume a claim is true just because it
       sounds convincing.
    5. Do not invent corrections or sources.
    6. If you cannot verify a claim, label it
       as "Needs verification".
    7. Return a corrected version of the report,
       followed by a short list of important corrections
       and claims that still need verification.

    Report:
    {report}
    """

    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )

    return response.text
```

```python
fact = fact_checker_agent(report)
print(fact)
```

```python
def orchestrator(task):
    task_lower = task.lower()

    if 'research' in task_lower:
        return research_agent(task)

    elif 'report' in task_lower or 'write' in task_lower:
        research = research_agent(task)

        report = writing_agent(research)

        checked_report = fact_checker_agent(report)

        return checked_report

    else:
        return "I don't know which agent to use"
```

```python
result = orchestrator(
    "Write a report about A Good Girl's Guide to Murder"
)
print(result)
```

error handling

```python
def research_agent(task):
    prompt = "Research this topic and explain it clearly: " + task

    try:
        response = client.models.generate_content(
            model="gemini-3.5-flash-lite",
            contents=prompt
        )

        if not response.text:
            return "Research failed: Gemini returned an empty response."

        return response.text

    except Exception as e:
        return f"Research failed: {e}"
```

```python
research_output = research_agent("Normal People by Sally Rooney")
print(research_output)
```

```python

def writing_agent(research_output):

    if research_output.startswith("Research failed:"):
        return research_output

    prompt = f"""
    You are a professional writer and editor.

    Your task is to turn the research below into a
    detailed, clear, well-structured report.

    Requirements:
    1. Create an informative title.
    2. Write an engaging introduction.
    3. Organize the content using headings and subheadings.
    4. Explain the important details clearly.
    5. Avoid repetition and irrelevant information.
    6. End with a meaningful conclusion.
    7. Do not invent facts that are not supported
       by the provided research.
    8. Use correct grammar and professional language.

    Research:
    {research_output}

    Write the final report.
    """

    try:
        response = client.models.generate_content(
            model="gemini-3.5-flash-lite",
            contents=prompt
        )

        if not response.text:
            return "Writing failed: Gemini returned an empty response."

        return response.text

    except Exception as e:
        return f"Writing failed: {e}"

```

```python
report = writing_agent(research_output)
print(report)
```

```python
def fact_checker_agent(report):

    prompt = f"""
    You are a fact-checking agent.

    Your task is to check the report below for factual accuracy.

    Instructions:
    1. Identify claims that may be factually incorrect.
    2. Identify information that is unsupported or uncertain.
    3. Suggest corrections wherever necessary.
    4. Do not rewrite the entire report.
    5. If you cannot verify a claim, clearly say so.

    Report:
    {report}

    Return your findings under these headings:
    - Potential factual errors
    - Unsupported or uncertain claims
    - Suggested corrections
    - Overall assessment
    """

    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=prompt
    )

    return response.text
```

```python
fact_check = fact_checker_agent(report)

print(fact_check)
```

```python
def orchestrator(task):
  print("Researching...")
  research=research_agent(task)
  print("Research Completed!")

  print("\n Writing...")
  report = writing_agent(research)
  print("Writing Completed!")

  print("\n Fact Checking...")
  fact_check = fact_checker_agent(report)
  print("Fact Checking Completed!")

  return {
      "research":research,
      "report":report,
      "fact_check":fact_check
  }
```

```python
result = orchestrator("Write a report about Crime and Punishment")

print("\n========== FINAL REPORT ==========\n")
print(result["report"])

print("\n========== FACT-CHECK RESULTS ==========\n")
print(result["fact_check"])
```

create a report revision agent

```python
def revision_agent(report, fact_check):

    prompt = f"""
    You are a report revision agent.

    Your task is to improve the report using the
    fact-checker's findings.

    ORIGINAL REPORT:
    {report}

    FACT-CHECK RESULTS:
    {fact_check}

    Instructions:
    1. Correct factual errors identified by the fact-checker.
    2. Incorporate valid suggested corrections.
    3. Preserve the original report's structure and useful analysis.
    4. Do not introduce new unsupported facts.
    5. Return only the complete revised report.
    """

    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )

    return response.text
```

```python
revised_report = revision_agent(
    result["report"],
    result["fact_check"]
)

print(revised_report)
```

```python
def orchestrator(task):

    print("Researching...")
    research = research_agent(task)
    print("Research Completed!")

    print("\nWriting...")
    report = writing_agent(research)
    print("Writing Completed!")

    print("\nFact Checking...")
    fact_check = fact_checker_agent(report)
    print("Fact Checking Completed!")

    print("\nRevising Report...")
    revised_report = revision_agent(report, fact_check)
    print("Revision Completed!")

    return {
        "research": research,
        "report": report,
        "fact_check": fact_check,
        "revised_report": revised_report
    }
```

add error handling to the orchestrator

```python

def orchestrator(task):

    try:
        print("Researching...")
        research = research_agent(task)
        print("Research Completed!")

        print("\nWriting...")
        report = writing_agent(research)
        print("Writing Completed!")

        print("\nFact Checking...")
        fact_check = fact_checker_agent(report)
        print("Fact Checking Completed!")

        print("\nRevising Report...")
        revised_report = revision_agent(report, fact_check)
        print("Revision Completed!")

        return {
            "research": research,
            "report": report,
            "fact_check": fact_check,
            "revised_report": revised_report
        }

    except Exception as e:
        print("Error occurred:", e)
        return None
```

```python
result = orchestrator("Write a report about The Kite Runner")

if result:
    print("\n========== FINAL REPORT ==========\n")
    print(result["report"])

    print("\n========== FACT-CHECK RESULTS ==========\n")
    print(result["fact_check"])

    print("\n========== REVISED REPORT ==========\n")
    print(result["revised_report"])
else:
    print("Orchestration failed")
```
